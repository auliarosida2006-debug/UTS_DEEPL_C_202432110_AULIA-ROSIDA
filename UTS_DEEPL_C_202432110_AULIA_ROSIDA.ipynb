```python
#202432110_AuliaRosida

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

print("Import library berhasil!")
print("PyTorch version:", torch.__version__)
print("CUDA tersedia:", torch.cuda.is_available())
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python

#202432110_AuliaRosida

# 1. Memuat dataset
df = pd.read_csv('/content/train.csv')

print("Ukuran dataset:", df.shape)
print("\nLima data pertama:")
display(df.head())

# 2. Mengubah kolom tanggal menjadi format datetime
df['date'] = pd.to_datetime(df['date'])

# 3. Memilih kategori produk GROCERY I
df = df[df['family'] == 'GROCERY I'].copy()

# 4. Menjumlahkan penjualan seluruh toko berdasarkan tanggal
data_harian = df.groupby('date')['sales'].sum().reset_index()

# 5. Mengurutkan data berdasarkan tanggal
data_harian = data_harian.sort_values('date')

# 6. Melihat karakteristik data
print("\nJumlah data setelah pemilihan kategori:",
      len(data_harian))

display(data_harian.head())

# 7. Membuat visualisasi data penjualan
plt.figure(figsize=(12, 5))
plt.plot(data_harian['date'], data_harian['sales'])
plt.title('Penjualan Harian Kategori GROCERY I')
plt.xlabel('Tanggal')
plt.ylabel('Jumlah Penjualan')
plt.grid(True)
plt.show()

# 8. Normalisasi data menggunakan MinMaxScaler
from sklearn.preprocessing import MinMaxScaler

values = data_harian['sales'].values.reshape(-1, 1)

# Membagi data secara kronologis 80:20
train_size = int(len(values) * 0.8)

scaler = MinMaxScaler()
scaler.fit(values[:train_size])

scaled_data = scaler.transform(values)

# 9. Membentuk sliding window
def create_sequences(data, time_step=30):
    X, y, target_indices = [], [], []

    for i in range(len(data) - time_step):
        X.append(data[i:i + time_step])
        y.append(data[i + time_step])
        target_indices.append(i + time_step)

    return (np.array(X), np.array(y),
            np.array(target_indices))

time_step = 30

X, y, target_indices = create_sequences(
    scaled_data, time_step
)

# 10. Membagi data training dan testing
train_mask = target_indices < train_size
test_mask = target_indices >= train_size

X_train = X[train_mask]
y_train = y[train_mask]

X_test = X[test_mask]
y_test = y[test_mask]

# 11. Konversi menjadi PyTorch Tensor
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

print("\nUkuran X_train:", X_train.shape)
print("Ukuran y_train:", y_train.shape)
print("Ukuran X_test:", X_test.shape)
print("Ukuran y_test:", y_test.shape)

print("\nPreprocessing berhasil!")
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python

#202432110_AuliaRosida

# Membuat arsitektur model LSTM
class LSTMModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64,
                 num_layers=2, output_size=1):
        super(LSTMModel, self).__init__()

        # Lapisan LSTM
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )

        # Lapisan output
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # Proses data melalui LSTM
        out, (hidden, cell) = self.lstm(x)

        # Mengambil output time step terakhir
        out = out[:, -1, :]

        # Menghasilkan prediksi
        out = self.fc(out)

        return out


# Inisialisasi model
model = LSTMModel()

# Fungsi loss
criterion = nn.MSELoss()

# Optimizer
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

# Menampilkan arsitektur model
print(model)
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>

</div>
